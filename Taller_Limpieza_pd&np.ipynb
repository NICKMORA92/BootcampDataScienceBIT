Parte 2 — Limpieza de datos (Pandas)
====================================
Realiza las siguientes tareas:

• Corregir tipos de datos incorrectos

• Identificar y manejar valores nulos

• Estandarizar columnas de texto:

• País

• Método de pago

• Revisar posibles valores atípicos (outliers)

• Corregir formatos de fecha si es necesario


```python
#Estandarizar tipos de datos
data['fecha'] = pd.to_datetime(data['fecha'], errors='coerce') 
data['cantidad'] = pd.to_numeric(data['cantidad'], errors='coerce')

# Verificar los tipos de datos después de la conversión
print(data.dtypes)
```

```python
# Manejo de nulos
data['cantidad'] = data['cantidad'].fillna(0)
data['precio'] = data['precio'].fillna(0)

print(data.isnull().sum())

display(data)
```

```python
#Estandarizar de columnas de texto
print("#"*100)
print(data['cliente'].unique())
print(data['producto'].unique())
print(data['pais'].unique())
print(data['metodo_pago'].unique())

print("#"*100)

data['metodo_pago'] = data['metodo_pago'].str.lower().str.strip()
print(data['metodo_pago'].unique())

data['pais'] = data['pais'].str.lower().str.strip()
print(data['pais'].unique())

#ajustar Col y perú
data['pais'] = data['pais'].replace({'col': 'colombia', 'perú': 'peru'})
print(data['pais'].unique())
```

```python
#normalizar precios = a 9999 con la media de precios
data['precio'] = data['precio'].replace(999999, data['precio'].mean())   
data['precio'] = data['precio'].round(2)
data.describe()
```

```python
print(data['fecha'].unique())
```

Respuestas:
-----------

¿Qué problemas encontraste?

R: Precio tenia valores incongruentes, pais tenia ortagrafía diferente para Perú y Colombia, los decimales del precio de Venta muy extensos

¿Cómo los solucionaste?

R: para el precio se reemplasa los valores de 99999 con la media de las ventas, en pais se cambia primero el campo todo a minúsculas y se reemplazan los valores errados, se redondean los decimales a máximo 2

¿Eliminaste registros? ¿Por qué?

R: los nulos, para poder dejar una base limpia y congruente.

Parte 3 — Análisis con Pandas
=============================
Actividades

• Crear una nueva columna llamada total: df['total'] = df['precio'] * df['cantidad']

Obtener:

• Total vendido

• Promedio de ventas

• Venta máxima

• Venta mínima

Identificar:

Los 5 productos con mayor valor vendido

El país con más ventas

```python
#nueva columna total
data['total'] = data['precio'] * data['cantidad']
data['total'] = data['total'].round(2)
display(data)


```

```python

# valores de resumen después de la limpieza

total_vendido = data['total'].sum().round(2)
promedio_ventas = data['total'].mean().round(2)
venta_mas_alta = data['total'].max().round(2)
venta_mas_baja = data['total'].min().round(2)


print(f'Total vendido: {total_vendido}')
print(f'Promedio de ventas: {promedio_ventas}')
print(f'Venta más alta: {venta_mas_alta}')
print(f'Venta más baja: {venta_mas_baja}')

productos_mas_vendidos = data.groupby('producto')['cantidad'].sum().sort_values(ascending=False).head(5)
print(f'Productos más vendidos:')
display(productos_mas_vendidos)

paises_mas_vendedores = data.groupby('pais')['total'].sum().sort_values(ascending=False).head(5)
print(f'Países con más ventas:')
display(paises_mas_vendedores)
```

Parte 4 — Introducción a NumPy
==============================
Actividades

• Convertir columnas numéricas a un array de NumPy: data = df[['precio',
'cantidad']].to_numpy()

• Separar columnas utilizando indexación: precios = data[:, 0]
cantidades = data[:, 1]

• Aplicar vectorización para calcular ventas totales: totales = precios *
cantidades

```python
# manejo de los datos con Numpy
data_np = data[['precio', 'cantidad']].to_numpy()

precios = data_np[:, 0]
cantidades = data_np[:, 1]

totales = precios * cantidades

print(totales[:5])
```

Parte 5 — Análisis con NumPy
============================
Utilizando funciones de NumPy, calcular:

• Suma total de ventas

• Promedio de ventas

• Venta máxima

• Cantidad de ventas superiores a 1000


```python
suma_totales = np.sum(totales)
print(f'Suma de totales: {suma_totales:.2f}')

promedio_ventas = np.mean(totales)
print(f'Promedio de ventas: {promedio_ventas:.2f}')

max_venta = np.max(totales)
print(f'Venta más alta: {max_venta:.2f}')

cantidad_mayor = np.sum(cantidades > 1000)
print(f'Cantidad de ventas con más de 1000 unidades: {cantidad_mayor}')
```

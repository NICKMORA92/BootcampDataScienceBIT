**Ejercicio 2 productos e inventario**

```python
productos_vendidos = [
    "Laptop", "Smartphone", "Tablet", "Laptop", "Smartwatch",
    "Headphones", "Camera", "Laptop", "Smartphone", "Tablet"]

inventario_disponible = [
    "Laptop", "Smartphone", "Tablet", "Smartwatch",
    "Headphones", "Camera", "Speaker", "Monitor", "Keyboard"]
```

```python
## todos los productos

total_productos = set(productos_vendidos).union(set(inventario_disponible))
print(total_productos)
```

```python
## productos vendidos en inventario

vendis_inventario = set(productos_vendidos).intersection(set(inventario_disponible))
print(vendis_inventario)
```

```python
## productos que no se han vendido

novendidos = set(inventario_disponible).difference(set(productos_vendidos))
print(novendidos)
```

```python
## evaluar producto

product = list(total_productos)

i = 0

while i < len(total_productos):
    productos = product[i]
    if productos in productos_vendidos:
        print(f"{productos} ha sido vendido.")
    else:
        print(f"{productos} no ha sido vendido.")
    i += 1
```

```python
#Explorar nulos
print(df.isnull().sum())
```

```python
filtered_df = df[df['Unnamed: 9'].notnull()]
display(filtered_df)



```

```python
#transformación de columnas en df
#Release-date
df['Release_Date'] = pd.to_datetime(df['Release_Date'], errors='coerce')
#Popularity
df['Popularity'] = pd.to_numeric(df['Popularity'], errors='coerce')
df['Vote_Count'] = pd.to_numeric(df['Vote_Count'], errors='coerce')
df['Vote_Average'] = pd.to_numeric(df['Vote_Average'], errors='coerce')

# --- Paso 1: VALIDAR qué filas se eliminarían de fechas vacias
filas_invalidas = df[df['Release_Date'].isna()]
print(f"Cantidad de filas con Release_Date inválida: {len(filas_invalidas)}")
print(f"Total de filas en df: {len(df)}")
print(f"Porcentaje a eliminar: {len(filas_invalidas)/len(df)*100:.2f}%")
display(filas_invalidas)

# --- Paso 2: ELIMINAR
df = df.dropna(subset=['Release_Date']).reset_index(drop=True)
print(f"Filas restantes: {len(df)}")
df.info()

# se eliminas las columnas Unnamed creadas por con error de valores corridos 
df = df.drop(columns=['Unnamed: 9', 'Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13'])
df.info()

```

```python
display(df)
```

```python
# Rellenar nulos de Popularity con la media de la columna
media_popularity = df['Popularity'].mean().round(2)
df['Popularity'] = df['Popularity'].fillna(media_popularity)

# Rellenar nulos de Vote_Count con la media de la columna
media_valC = df['Vote_Count'].mean().round(2)
df['Vote_Count'] = df['Vote_Count'].fillna(media_valC)

# Rellenar nulos de Vote_Average con la media de la columna
media_valA = df['Vote_Average'].mean().round(2)
df['Vote_Average'] = df['Vote_Average'].fillna(media_valA)

display(df.head(20))
```

```python
df.info()

print('='* 40)

print(df.isnull().sum())
```

```python
# Rellenar nulos en columnas de texto con "Not asigned"
cols_texto = ['Overview', 'Original_Language', 'Genre', 'Poster_Url']

for col in cols_texto:
    print(f"{col} → nulos antes: {df[col].isnull().sum()}")
    df[col] = df[col].fillna('Not asigned')
    print(f"{col} → nulos después: {df[col].isnull().sum()}")

display(df.head(20))
```

```python
df.info()

print('='* 40)

print(df.isnull().sum())
```

```python
# Exploración de datos usando pandas

df_leng = pd.DataFrame(df['Original_Language'].value_counts())
df_leng.columns = ['Cantidad']
df_leng['Porcentaje'] = (df_leng['Cantidad'] / df_leng['Cantidad'].sum() * 100).round(2)
df_leng['Porcentaje'] = df_leng['Porcentaje'].astype(str) + ' %'
display(df_leng)
```

```python
df_tipo = pd.DataFrame(df['Genre'])

# split devuelve una lista por celda
df_tipo['Genre_list'] = df_tipo['Genre'].str.split(',')

# detectar máximo de géneros en una fila
max_genres = df_tipo['Genre_list'].str.len().max()
print(f"Máximo de géneros por fila: {max_genres}")

# crear Genre_1, Genre_2, ... dinámicamente
for i in range(max_genres):
    df_tipo[f'Genre_{i+1}'] = df_tipo['Genre_list'].str[i]

# limpiar nulos en las nuevas columnas
df_tipo[[f'Genre_{i+1}' for i in range(max_genres)]] = \
    df_tipo[[f'Genre_{i+1}' for i in range(max_genres)]].fillna('Not asigned')

df_tipo = df_tipo.drop(columns=['Genre_list'])
display(df_tipo)
```

```python
import matplotlib.pyplot as plt

# Conteo por género
genre_counts = df_tipo['Genre_1'].value_counts()

plt.figure(figsize=(12, 6))
bars = plt.bar(genre_counts.index, genre_counts.values, color='steelblue', edgecolor='black')

# Valores encima de cada barra
for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, h + 5, f'{int(h)}',
             ha='center', va='bottom', fontsize=9)

plt.title('Conteo de películas por Género principal (Genre_1)', fontsize=14, fontweight='bold')
plt.xlabel('Género')
plt.ylabel('Cantidad de películas')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Media de votos por género
genre_votes = df.groupby('Title')['Vote_Average'].mean().sort_values(ascending=False)
display(genre_votes.head(10))
```

```python
from IPython.display import display, Markdown

df_cross = df[['Title', 'Popularity']].join(df_tipo['Genre_1'])
df_cross = df_cross[df_cross['Genre_1'] != 'Not asigned']

for genero, grupo in (
    df_cross
    .sort_values('Popularity', ascending=False)
    .groupby('Genre_1')
):
    top5 = grupo.head(5)
    md = f"### 🎬 {genero}\n\n"
    md += "| # | Título | Popularity |\n"
    md += "|---|--------|-----------|\n"
    for i, (_, row) in enumerate(top5.iterrows(), 1):
        md += f"| {i} | {row['Title']} | {row['Popularity']:.2f} |\n"
    display(Markdown(md))
```

```python
# fechas de lanzamiento

df['Release_Year'] = df['Release_Date'].dt.year
plt.figure(figsize=(12, 6))
sns.histplot(df['Release_Year'], bins=30, kde=False, color='cornflowerblue', edgecolor='black')
plt.title('Distribución de Películas por Año de Lanzamiento', fontsize=14, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Cantidad de Películas')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
